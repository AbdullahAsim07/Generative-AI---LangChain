UnStructuredURlLoader

```python
# url dobara load na hoo is liye
import os
os.environ["UNSTRUCTURED_DISABLE_NOTICE"] = "1"
# internet sa data load karne ka liye hai unstructured url use karte hai
from langchain_community.document_loaders import UnstructuredURLLoader
loader=UnstructuredURLLoader(
    urls=["https://www.cnbc.com/2024/12/21/how-the-federal-reserves-rate-policy-affects-mortgages.html"]
)
# yeh url ko load liya sath aik varaible ma save kiya 
docs=loader.load()
# yeh url ki len check karne ka liye
len(docs)
```

```python
# hum is url ka metadata check kiya jis ma loaction,name, image ka pata chalte hai etc 
docs[0].metadata
```

```python
docs[0].page_content
```
