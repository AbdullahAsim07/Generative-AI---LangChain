CHARACTER TEXT SPLITTER

```python
# langchain kehta hai tumhe python ki tarah slice aur for loop lagane ki zarroat nhi ma tumhe builtin library deta hoo tum bas yeh line likho
from langchain_text_splitters import CharacterTextSplitter
# Yeh bina new line (\n) ke bhi text ko smartly chote chunks mein tod deta hai
splitter = CharacterTextSplitter(
    separator="/n",
    chunk_size=300,
    chunk_overlap=0
)


```

```python
chunks = splitter.split_text(text)
len(chunks)
```

```python
# Ab yeh error nahi dega kyunki chunks[1] ab maujood hoga
len(chunks[0]),len(chunks[1])
```

RECURSIVE TEXT SPLITTER

```python
# Multiple line breaks aur separators configure karne ke liye
from langchain_text_splitters import RecursiveCharacterTextSplitter

r_splitter = RecursiveCharacterTextSplitter(
    # Sahi backslash (\n) lagayein, forward slash (/n) kaam nahi karega
    separators=["\n\n", "\n", " ", ""],
    
    chunk_size=300,
    # S' ko hata kar sirf 'chunk_overlap' likhein
    chunk_overlap=20
)

# Chunks banane ke liye
chunks = r_splitter.split_text(text)
print("Total Chunks:", len(chunks))
```

```python
# ab hum is par for loop lagaye ga 
chunks=r_splitter.split_text(text)
for chunk in chunks:
 print(len(chunks))
```
